```python
import pandas as pd
from pathlib import Path

RAW = Path(r"C:\Users\nagar\Downloads\routeranker-upload\data\raw\Banglore")

files = {
    "trips": RAW / "trips.txt",
    "stops": RAW / "stops.txt",
    "stop_times": RAW / "stop_times.txt"
}

# Load files
trips = pd.read_csv(files["trips"], dtype=str)
stops = pd.read_csv(files["stops"], dtype=str)
stop_times = pd.read_csv(files["stop_times"], dtype=str)

# Remove accidental spaces from column names
for df in [trips, stops, stop_times]:
    df.columns = df.columns.str.strip()

print("BANGALORE GTFS VALIDATION")
print("=" * 50)

print("Trips rows:", len(trips))
print("Stops rows:", len(stops))
print("Stop times rows:", len(stop_times))

print("\nRequired columns:")
print("trips.txt:", "trip_id" in trips.columns)
print("stops.txt:", "stop_id" in stops.columns)
print("stop_times.txt:", all(
    col in stop_times.columns
    for col in ["trip_id", "stop_id", "arrival_time",
                "departure_time", "stop_sequence"]
))

# Check matching IDs
trip_ids = set(trips["trip_id"].dropna().str.strip())
stop_trip_ids = set(stop_times["trip_id"].dropna().str.strip())

stop_ids = set(stops["stop_id"].dropna().str.strip())
used_stop_ids = set(stop_times["stop_id"].dropna().str.strip())

print("\nTRIP ID VALIDATION")
print("Matching trip IDs:", len(trip_ids & stop_trip_ids))
print("Unmatched stop-time trip IDs:", len(stop_trip_ids - trip_ids))

print("\nSTOP ID VALIDATION")
print("Matching stop IDs:", len(stop_ids & used_stop_ids))
print("Unmatched stop-time stop IDs:", len(used_stop_ids - stop_ids))
```

```python
from pathlib import Path
import shutil
from datetime import datetime

RAW = Path(
    r"C:\Users\nagar\Downloads\routeranker-upload\data\raw\Banglore"
)

DEST = Path(
    r"C:\Users\nagar\Downloads\routeranker-upload\notebooks\gtfs_data\Bangalore"
)

DEST.mkdir(parents=True, exist_ok=True)

files = [
    "routes.txt",
    "trips.txt",
    "stops.txt",
    "stop_times.txt",
    "calendar.txt"
]

# Create backup folder
backup = DEST / ("backup_" + datetime.now().strftime("%Y%m%d_%H%M%S"))

for filename in files:
    source = RAW / filename
    target = DEST / filename

    if not source.exists():
        print("SOURCE MISSING:", filename)
        continue

    if target.exists():
        backup.mkdir(exist_ok=True)
        shutil.copy2(target, backup / filename)

    shutil.copy2(source, target)

    print("COPIED SUCCESSFULLY:", filename)

print("\nBackup folder:", backup)
print("Bangalore GTFS replacement completed.")
```

```python

from pathlib import Path
from zipfile import ZipFile
from io import BytesIO
import pandas as pd

# Update this path to your original GTFS ZIP folder
ZIP_FOLDER = PROJECT_DIR / "original_downloads"

# Each city has its own original GTFS ZIP file
zip_files = {
    "Hyderabad": ZIP_FOLDER / "Hyderabad.zip",
    "Bangalore": ZIP_FOLDER / "Bangalore.zip",
    "Delhi": ZIP_FOLDER / "Delhi.zip",
    "Mumbai": ZIP_FOLDER / "Mumbai.zip",
    "Pune": ZIP_FOLDER / "Pune.zip"
}

required_columns = [
    "trip_id",
    "arrival_time",
    "departure_time",
    "stop_id",
    "stop_sequence"
]

for city, zip_path in zip_files.items():

    output_folder = PROJECT_DIR / "data" / "raw" / city
    output_folder.mkdir(parents=True, exist_ok=True)

    output_file = output_folder / "stop_times.txt"

    print(f"\nProcessing {city}...")

    if not zip_path.exists():
        print("ZIP file not found:", zip_path)
        continue

    with ZipFile(zip_path, "r") as archive:

        # Find stop_times.txt even inside nested folders
        matches = [
            name for name in archive.namelist()
            if Path(name).name.lower() == "stop_times.txt"
        ]

        if not matches:
            print("stop_times.txt not found in ZIP")
            continue

        if len(matches) > 1:
            print("Multiple stop_times.txt files found.")
            print(matches)
            print("Please select the correct feed manually.")
            continue

        # Read the original GTFS file using Pandas
        with archive.open(matches[0]) as file:
            df = pd.read_csv(
                BytesIO(file.read()),
                dtype={"trip_id": str, "stop_id": str}
            )

        # Check required columns
        missing_columns = [
            col for col in required_columns
            if col not in df.columns
        ]

        if missing_columns:
            print("Missing columns:", missing_columns)
            continue

        # Save the original records in the correct city folder
        df.to_csv(output_file, index=False)

        print("Saved successfully:", output_file)
        print("Total records:", len(df))
        print("Total columns:", len(df.columns))
        print(df.head(3))
```

```python

import shutil
from pathlib import Path

project = Path.cwd().parent
raw_folder = project / "data" / "raw"

cities = ["Hyderabad", "Bangalore", "Delhi", "Mumbai", "Pune"]

for city in cities:

    destination_folder = raw_folder / city
    destination_folder.mkdir(parents=True, exist_ok=True)

    matches = [
        file for file in project.rglob("stop_times.txt")
        if city.lower() in str(file).lower()
        and raw_folder not in file.parents
    ]

    if len(matches) == 1:
        destination = destination_folder / "stop_times.txt"
        shutil.copy2(matches[0], destination)
        print(f"SUCCESS: {city} copied")

    elif len(matches) == 0:
        print(f"NOT FOUND: {city}")

    else:
        print(f"MULTIPLE MATCHES: {city}")
        for file in matches:
            print(file)
```

```python

import pandas as pd
from pathlib import Path

# 1. Main project data folder
BASE_DIR = Path("data/raw")

# 2. All five cities
cities = [
    "Banglore",
    "Hyderabad",
    "Delhi",
    "Mumbai",
    "Pune"
]

# 3. Required GTFS files and their important columns
required_columns = {
    "routes.txt": ["route_id"],
    "trips.txt": ["route_id", "trip_id", "service_id"],
    "stop_times.txt": [
        "trip_id", "arrival_time",
        "departure_time", "stop_id", "stop_sequence"
    ],
    "stops.txt": ["stop_id", "stop_name"]
}

results = []

print("CURRENT WORKING DIRECTORY:", Path.cwd())
print("\nSTARTING ALL-CITY GTFS VERIFICATION")

for city in cities:

    city_path = BASE_DIR / city

    print("\n" + "=" * 60)
    print("CITY:", city)
    print("=" * 60)

    if not city_path.exists():
        print("ERROR: City folder not found!")

        results.append({
            "City": city,
            "File": "City folder",
            "Records": 0,
            "Status": "MISSING"
        })
        continue

    for filename, expected_columns in required_columns.items():

        file_path = city_path / filename

        if not file_path.exists():
            print(filename, "MISSING")

            results.append({
                "City": city,
                "File": filename,
                "Records": 0,
                "Status": "MISSING"
            })
            continue

        try:
            df = pd.read_csv(file_path, low_memory=False)

            missing_columns = [
                col for col in expected_columns
                if col not in df.columns
            ]

            if len(df) == 0:
                status = "EMPTY"
            elif missing_columns:
                status = "CHECK COLUMNS"
            else:
                status = "LOADED"

            print("\nFile:", filename)
            print("Records:", len(df))
            print("Size:", round(file_path.stat().st_size / (1024*1024), 2), "MB")
            print("Missing values:", int(df.isnull().sum().sum()))
            print("Duplicate rows:", int(df.duplicated().sum()))
            print("Columns:", list(df.columns))
            print("Status:", status)

            if missing_columns:
                print("Missing required columns:", missing_columns)

            results.append({
                "City": city,
                "File": filename,
                "Records": len(df),
                "Missing Values": int(df.isnull().sum().sum()),
                "Duplicate Rows": int(df.duplicated().sum()),
                "Status": status
            })

            del df

        except Exception as e:
            print("ERROR:", str(e))

            results.append({
                "City": city,
                "File": filename,
                "Records": 0,
                "Status": "READ ERROR"
            })

    # Calendar or calendar_dates check
    calendar_found = False

    for filename in ["calendar.txt", "calendar_dates.txt"]:
        file_path = city_path / filename

        if file_path.exists():
            calendar_found = True

            try:
                cal_df = pd.read_csv(file_path)

                status = "LOADED" if len(cal_df) > 0 else "EMPTY"

                print("\n", filename, ":", status)
                print("Records:", len(cal_df))
                print("Columns:", list(cal_df.columns))

                results.append({
                    "City": city,
                    "File": filename,
                    "Records": len(cal_df),
                    "Status": status
                })

            except Exception as e:
                print(filename, "READ ERROR:", e)

                results.append({
                    "City": city,
                    "File": filename,
                    "Records": 0,
                    "Status": "READ ERROR"
                })

            del cal_df

    if not calendar_found:
        print("\nWARNING: No calendar.txt or calendar_dates.txt found")

        results.append({
            "City": city,
            "File": "Calendar files",
            "Records": 0,
            "Status": "MISSING"
        })

# 4. Final verification report
inventory = pd.DataFrame(results)

print("\n" + "=" * 60)
print("FINAL ALL-CITY DATASET VERIFICATION")
print("=" * 60)

display(inventory)

# 5. Save verification report
Path("reports").mkdir(exist_ok=True)

inventory.to_csv(
    "reports/module1_final_verification.csv",
    index=False
)

print("\nVerification report saved successfully!")
print("Location: reports/module1_final_verification.csv")

# 6. Overall status
if (inventory["Status"] == "LOADED").all():
    print("\nALL CHECKED FILES LOADED SUCCESSFULLY")
else:
    print("\nSOME FILES NEED ATTENTION. CHECK THE REPORT ABOVE.")
```
