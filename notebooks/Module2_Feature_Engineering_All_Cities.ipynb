```python
# ==========================================
# STEP 11.1 - TIME FEATURE EXTRACTION
# ==========================================

import re

def parse_gtfs_time(value):

    if pd.isna(value):
        return np.nan

    parts = str(value).strip().split(":")

    if len(parts) != 3:
        return np.nan

    try:
        hour, minute, second = map(int, parts)

        if hour < 0 or not (0 <= minute < 60) or not (0 <= second < 60):
            return np.nan

        return hour * 3600 + minute * 60 + second

    except (ValueError, TypeError):
        return np.nan


def classify_time(hour):

    if pd.isna(hour):
        return "Unknown"

    if 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"


# Create output folder
TIME_OUTPUT = OUTPUT_PATH / "time_features"
TIME_OUTPUT.mkdir(parents=True, exist_ok=True)

TIME_FEATURES = {}

for city in CITY_PATHS:

    print("\nProcessing time features:", city)

    stop_times = GTFS_DATA[city]["stop_times"].copy()

    # Parse departure and arrival times separately
    departure = stop_times["departure_time"].map(parse_gtfs_time)

    arrival = stop_times["arrival_time"].map(parse_gtfs_time)

    # Use arrival time when departure time is unavailable or invalid
    stop_times["scheduled_seconds"] = departure.combine_first(arrival)

    # Identify the time source
    stop_times["time_source"] = np.where(
        departure.notna(),
        "departure_time",
        np.where(arrival.notna(), "arrival_time", "missing")
    )

    # Extract service hour and minute
    stop_times["service_hour"] = (
        stop_times["scheduled_seconds"] // 3600
    )

    stop_times["minute"] = (
        (stop_times["scheduled_seconds"] // 60) % 60
    )

    # Convert extended GTFS time to ordinary clock hour
    stop_times["departure_hour"] = (
        stop_times["service_hour"] % 24
    )

    # Classify time period using clock hour
    stop_times["time_period"] = (
        stop_times["departure_hour"].apply(classify_time)
    )

    stop_times["city"] = city

    TIME_FEATURES[city] = stop_times

    # Save city time features
    output_file = TIME_OUTPUT / f"{city.upper()}_time_features.csv"

    stop_times.to_csv(output_file, index=False)

    print("Total records:", len(stop_times))
    print("Valid time records:", stop_times["scheduled_seconds"].notna().sum())
    print("Invalid or missing times:", stop_times["scheduled_seconds"].isna().sum())
    print("Saved:", output_file)

print("\nTIME FEATURE EXTRACTION COMPLETED")
```

```python
# ==========================================
# STEP 11.2 - HOURLY TRIP COUNTS
# ==========================================

HOURLY_OUTPUT = OUTPUT_PATH / "hourly_counts"
HOURLY_OUTPUT.mkdir(parents=True, exist_ok=True)

HOURLY_COUNTS = {}

for city in CITY_PATHS:

    print("\nProcessing hourly counts:", city)

    df = TIME_FEATURES[city].copy()

    # Sort by stop sequence
    df["stop_sequence"] = pd.to_numeric(
        df["stop_sequence"],
        errors="coerce"
    )

    df = df.sort_values(
        ["trip_id", "stop_sequence"]
    )

    # Keep first valid scheduled time per trip
    first_stops = (
        df.dropna(subset=["trip_id", "scheduled_seconds"])
          .drop_duplicates(subset=["trip_id"], keep="first")
          .copy()
    )

    # Add route ID using trips.txt
    trips = GTFS_DATA[city]["trips"]

    trip_routes = trips[
        ["trip_id", "route_id"]
    ].drop_duplicates(subset=["trip_id"])

    first_stops = first_stops.merge(
        trip_routes,
        on="trip_id",
        how="left"
    )

    # Calculate distinct scheduled trips per route and hour
    hourly = (
        first_stops
        .dropna(subset=["route_id", "service_hour"])
        .groupby(
            ["route_id", "service_hour", "departure_hour", "time_period"],
            as_index=False
        )
        .agg(
            trip_count=("trip_id", "nunique")
        )
    )

    hourly["city"] = city

    HOURLY_COUNTS[city] = hourly

    output_file = (
        HOURLY_OUTPUT / f"{city.upper()}_hourly_counts.csv"
    )

    hourly.to_csv(output_file, index=False)

    print("Routes with scheduled trips:", hourly["route_id"].nunique())
    print("Hourly records:", len(hourly))
    print("Total distinct trips counted:", first_stops["trip_id"].nunique())
    print("Saved:", output_file)

print("\nALL FIVE CITIES HOURLY COUNTS COMPLETED")
```

```python
# ==========================================
# STEP 12.1 - ROUTE TRIP AND STOP COUNTS
# ==========================================

ROUTE_OUTPUT = OUTPUT_PATH / "route_features"
ROUTE_OUTPUT.mkdir(parents=True, exist_ok=True)

ROUTE_FEATURES = {}

for city in CITY_PATHS:

    print("\n================================")
    print("PROCESSING:", city)
    print("================================")

    # Load required DataFrames
    routes = GTFS_DATA[city]["routes"].copy()
    trips = GTFS_DATA[city]["trips"].copy()
    stop_times = GTFS_DATA[city]["stop_times"].copy()

    # --------------------------------------
    # 1. Calculate distinct trips per route
    # --------------------------------------

    trip_mapping = trips[
        ["route_id", "trip_id"]
    ].dropna().drop_duplicates()

    trip_counts = (
        trip_mapping
        .groupby("route_id")["trip_id"]
        .nunique()
        .reset_index(name="trip_count")
    )

    # --------------------------------------
    # 2. Calculate distinct stops per route
    # --------------------------------------

    stop_mapping = (
        stop_times[["trip_id", "stop_id"]]
        .dropna()
        .drop_duplicates()
        .merge(
            trip_mapping,
            on="trip_id",
            how="inner"
        )
    )

    stop_counts = (
        stop_mapping
        .groupby("route_id")["stop_id"]
        .nunique()
        .reset_index(name="stop_count")
    )

    # --------------------------------------
    # 3. Combine route information
    # --------------------------------------

    route_profile = routes.drop_duplicates(
        subset=["route_id"]
    ).copy()

    route_profile = route_profile.merge(
        trip_counts,
        on="route_id",
        how="left"
    )

    route_profile = route_profile.merge(
        stop_counts,
        on="route_id",
        how="left"
    )

    # Zero means no corresponding records in
    # the available GTFS source tables.
    route_profile["trip_count"] = (
        route_profile["trip_count"].fillna(0).astype(int)
    )

    route_profile["stop_count"] = (
        route_profile["stop_count"].fillna(0).astype(int)
    )

    route_profile["city"] = city

    # --------------------------------------
    # 4. Save route profile
    # --------------------------------------

    ROUTE_FEATURES[city] = route_profile

    output_file = (
        ROUTE_OUTPUT / f"{city.upper()}_route_features.csv"
    )

    route_profile.to_csv(output_file, index=False)

    print("Total routes:", len(route_profile))
    print("Routes with trips:", (route_profile["trip_count"] > 0).sum())
    print("Total distinct route-trip pairs:", trip_mapping["trip_id"].nunique())
    print("Routes with stops:", (route_profile["stop_count"] > 0).sum())
    print("Saved:", output_file)

print("\nALL FIVE CITIES ROUTE FEATURES COMPLETED")
```

```python
# STEP 13: COMBINE ROUTE FEATURES AND HOURLY COUNTS

ALL_CITY_DATASETS = []

for city in CITY_PATHS:

    print("\nProcessing:", city)

    route_df = ROUTE_FEATURES[city].copy()
    hourly_df = HOURLY_COUNTS[city].copy()

    # Calculate hourly service summaries for each route
    hourly_summary = (
        hourly_df.groupby("route_id")
        .agg(
            hourly_trip_total=("trip_count", "sum"),
            peak_service_hour=("service_hour",
                               lambda x: x.mode().iloc[0]
                               if not x.mode().empty else np.nan),
            peak_hour_trip_count=("trip_count", "max")
        )
        .reset_index()
    )

    # Merge hourly summaries with route features
    combined = route_df.merge(
        hourly_summary,
        on="route_id",
        how="left"
    )

    # Fill missing hourly values with zero
    combined["hourly_trip_total"] = (
        combined["hourly_trip_total"].fillna(0).astype(int)
    )

    combined["peak_hour_trip_count"] = (
        combined["peak_hour_trip_count"].fillna(0).astype(int)
    )

    combined["city"] = city

    ALL_CITY_DATASETS.append(combined)

    print("Routes:", len(combined))
    print("Columns:", list(combined.columns))

# Combine all five cities
all_cities_route_dataset = pd.concat(
    ALL_CITY_DATASETS,
    ignore_index=True,
    sort=False
)

# Create output folder
DATASET_OUTPUT = OUTPUT_PATH / "route_dataset"
DATASET_OUTPUT.mkdir(parents=True, exist_ok=True)

# Save combined dataset
dataset_file = DATASET_OUTPUT / "all_cities_route_dataset.csv"

all_cities_route_dataset.to_csv(dataset_file, index=False)

print("\nALL FIVE CITIES COMBINED SUCCESSFULLY")
print("Total records:", len(all_cities_route_dataset))
print("Saved:", dataset_file)
```

```python
gtfs_data = {}

required_files = [
    "routes.txt",
    "trips.txt",
    "stops.txt",
    "stop_times.txt",
    "calendar.txt"
]

optional_files = [
    "calendar_dates.txt",
    "shapes.txt"
]

for city, folder in city_folders.items():

    print("\n" + "=" * 50)
    print("CITY:", city)
    print("=" * 50)

    city_path = GTFS_ROOT / folder
    gtfs_data[city] = {}

    for filename in required_files + optional_files:

        file_path = city_path / filename

        if not file_path.exists():
            print(filename, "- NOT FOUND")
            continue

        if file_path.stat().st_size == 0:
            print(filename, "- EMPTY FILE")
            continue

        try:
            df = pd.read_csv(file_path, low_memory=False)

            gtfs_data[city][filename.replace(".txt", "")] = df

            print(
                filename,
                "| Rows:", len(df),
                "| Columns:", len(df.columns),
                "| LOADED"
            )

        except EmptyDataError:
            print(filename, "- EMPTY DATA")

        except Exception as e:
            print(filename, "- ERROR:", e)
```

```python
hourly_results = []

for city, files in gtfs_data.items():

    trips = files["trips"]
    stop_times = files["stop_times"]

    # Join trips with their route IDs
    merged = stop_times.merge(
        trips[["trip_id", "route_id"]],
        on="trip_id",
        how="inner"
    )

    # Sort stops to identify the first stop of each trip
    merged["stop_sequence"] = pd.to_numeric(
        merged["stop_sequence"], errors="coerce"
    )

    merged = merged.sort_values("stop_sequence")

    first_departures = merged.drop_duplicates(
        subset="trip_id",
        keep="first"
    ).copy()

    # Extract scheduled departure hour
    time_parts = first_departures["departure_time"].astype(
        "string"
    ).str.extract(r"^\s*(\d+):([0-5]\d):([0-5]\d)\s*$")

    first_departures["hour"] = pd.to_numeric(
        time_parts[0], errors="coerce"
    )

    first_departures = first_departures.dropna(subset=["hour"])

    first_departures["hour"] = (
        first_departures["hour"].astype(int) % 24
    )

    # Count trips per hour
    hourly = (
        first_departures.groupby("hour")["trip_id"]
        .nunique()
        .reset_index(name="trip_count")
    )

    hourly["city"] = city

    hourly_results.append(hourly)

    print(city, "- hourly counts generated")

hourly_counts = pd.concat(hourly_results, ignore_index=True)

hourly_counts = hourly_counts[
    ["city", "hour", "trip_count"]
]

display(hourly_counts.head(20))
```

```python
import pandas as pd

route_results = []

for city, files in gtfs_data.items():

    print("\nProcessing:", city)

    routes = files["routes"].copy()
    trips = files["trips"].copy()
    stop_times = files["stop_times"].copy()

    # 1. Trip count per route
    trip_features = (
        trips.groupby("route_id")
        .agg(
            trip_count=("trip_id", "nunique"),
            service_count=("service_id", "nunique")
        )
        .reset_index()
    )

    # 2. Join trips with stop_times
    trip_stops = trips[["route_id", "trip_id"]].merge(
        stop_times[["trip_id", "stop_id"]],
        on="trip_id",
        how="inner"
    )

    # 3. Distinct stop count per route
    stop_features = (
        trip_stops.groupby("route_id")
        .agg(
            distinct_stop_count=("stop_id", "nunique")
        )
        .reset_index()
    )

    # 4. Average stops per trip
    stops_per_trip = (
        trip_stops.groupby(["route_id", "trip_id"])
        ["stop_id"].nunique()
        .reset_index(name="stops_per_trip")
    )

    average_stops = (
        stops_per_trip.groupby("route_id")
        ["stops_per_trip"].mean()
        .reset_index(name="average_stops_per_trip")
    )

    # 5. Combine route features
    features = routes.merge(
        trip_features, on="route_id", how="left"
    )

    features = features.merge(
        stop_features, on="route_id", how="left"
    )

    features = features.merge(
        average_stops, on="route_id", how="left"
    )

    features["city"] = city

    route_results.append(features)

    print("Routes processed:", len(features))

# Combine all cities
all_route_features = pd.concat(
    route_results,
    ignore_index=True
)

print("\nAll five cities processed!")
print("Total route records:", len(all_route_features))

display(all_route_features.head())
```

```python
import pandas as pd

time_results = []

for city, files in gtfs_data.items():

    print("\nProcessing:", city)

    stop_times = files["stop_times"].copy()

    # Extract departure time components
    parts = stop_times["departure_time"].astype(
        "string"
    ).str.extract(r"^\s*(\d+):([0-5]\d):([0-5]\d)\s*$")

    stop_times["departure_hour"] = pd.to_numeric(
        parts[0], errors="coerce"
    )

    stop_times["departure_minute"] = pd.to_numeric(
        parts[1], errors="coerce"
    )

    # Clock hour (handles GTFS times above 24:00)
    stop_times["clock_hour"] = (
        stop_times["departure_hour"] % 24
    )

    # Categorize time periods
    def get_period(hour):
        if pd.isna(hour):
            return "Unknown"
        elif 5 <= hour < 12:
            return "Morning"
        elif 12 <= hour < 17:
            return "Afternoon"
        elif 17 <= hour < 21:
            return "Evening"
        else:
            return "Night"

    stop_times["time_period"] = (
        stop_times["clock_hour"].apply(get_period)
    )

    stop_times["city"] = city

    selected = stop_times[
        [
            "trip_id",
            "stop_id",
            "departure_time",
            "departure_hour",
            "departure_minute",
            "clock_hour",
            "time_period"
        ]
    ].copy()

    selected["city"] = city

    time_results.append(selected)

    print("Time features generated:", len(selected))

# Combine all five cities
all_time_features = pd.concat(
    time_results,
    ignore_index=True
)

print("\nAll city time features generated!")
display(all_time_features.head())
```

```python
import pandas as pd

headway_results = []

for city, files in gtfs_data.items():

    trips = files["trips"].copy()
    stop_times = files["stop_times"].copy()

    # Select the first stop of every trip
    stop_times["stop_sequence"] = pd.to_numeric(
        stop_times["stop_sequence"], errors="coerce"
    )

    first_stops = (
        stop_times
        .sort_values("stop_sequence")
        .drop_duplicates(subset="trip_id", keep="first")
    )

    # Include route, service and direction information
    trip_columns = ["trip_id", "route_id", "service_id"]

    if "direction_id" in trips.columns:
        trip_columns.append("direction_id")

    merged = first_stops.merge(
        trips[trip_columns],
        on="trip_id",
        how="inner"
    )

    # Convert departure time into minutes from midnight
    time_parts = merged["departure_time"].astype("string").str.extract(
        r"^\s*(\d+):([0-5]\d):([0-5]\d)\s*$"
    )

    merged["hour"] = pd.to_numeric(time_parts[0], errors="coerce")
    merged["minute"] = pd.to_numeric(time_parts[1], errors="coerce")

    merged = merged.dropna(subset=["hour", "minute"])

    merged["departure_minutes"] = (
        merged["hour"] * 60 + merged["minute"]
    )

    # Group trips by route, service and direction
    group_columns = ["route_id", "service_id"]

    if "direction_id" in merged.columns:
        group_columns.append("direction_id")

    merged = merged.sort_values("departure_minutes")

    merged["headway_minutes"] = (
        merged.groupby(group_columns)["departure_minutes"]
        .diff()
    )

    # Keep valid positive headways
    valid = merged[merged["headway_minutes"] > 0].copy()

    valid["city"] = city
    headway_results.append(valid)

all_cities_headways = pd.concat(
    headway_results,
    ignore_index=True
)

display(
    all_cities_headways[
        ["city", "route_id", "service_id",
         "departure_time", "headway_minutes"]
    ].head(10)
)

print("Total headway records:", len(all_cities_headways))
```

```python
# Sort trips by scheduled departure time within each group
group_cols = ["city", "route_id", "service_id", "direction_id"]

headway_check = headway.copy()

headway_check = headway_check.sort_values(
    group_cols + ["departure_minutes"]
)

# Recalculate the gap from the previous trip
headway_check["expected_headway"] = (
    headway_check.groupby(
        group_cols,
        dropna=False
    )["departure_minutes"].diff()
)

print("Expected missing headways:",
      headway_check["expected_headway"].isna().sum())

print("Existing missing headways:",
      headway_check["headway_minutes"].isna().sum())

print("Mismatched headway values:",
      (
          headway_check["headway_minutes"]
          != headway_check["expected_headway"]
      ).sum())

print("\nSample recalculated values:")
display(
    headway_check[
        ["city", "route_id", "direction_id",
         "departure_minutes", "headway_minutes",
         "expected_headway"]
    ].head(15)
)
```

```python
# Replace incorrect headways with recalculated values
headway_check["headway_minutes"] = headway_check["expected_headway"]

# Remove temporary validation column
headway_final = headway_check.drop(
    columns=["expected_headway"]
)

print("Corrected headway records:", len(headway_final))
print("Missing headways:", headway_final["headway_minutes"].isna().sum())
print("Negative headways:", (headway_final["headway_minutes"] < 0).sum())
print("Zero headways:", (headway_final["headway_minutes"] == 0).sum())
```
