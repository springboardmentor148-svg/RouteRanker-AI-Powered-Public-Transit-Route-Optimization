```python
from pathlib import Path
import pandas as pd

raw_folder = Path("../data/raw/Hyderabad")

# Load original stop-time records
original_stop_times = pd.read_csv(
    raw_folder / "stop_times.txt",
    usecols=[
        "trip_id",
        "arrival_time",
        "departure_time",
        "stop_id",
        "stop_sequence"
    ],
    dtype=str,
    low_memory=False
)

# Load original trip information
original_trips = pd.read_csv(
    raw_folder / "trips.txt",
    usecols=[
        "trip_id",
        "route_id",
        "service_id"
    ],
    dtype=str,
    low_memory=False
)

# Inspect the trip that had departure time 86:05:00
trip_ids_to_check = ["41854447", "41755961"]

for trip_id in trip_ids_to_check:
    print(f"\n========== TRIP {trip_id} ==========")

    trip_details = original_trips[
        original_trips["trip_id"] == trip_id
    ]

    print("\nTrip information:")
    display(trip_details)

    trip_stops = original_stop_times[
        original_stop_times["trip_id"] == trip_id
    ].copy()

    trip_stops["stop_sequence_num"] = pd.to_numeric(
        trip_stops["stop_sequence"],
        errors="coerce"
    )

    trip_stops = trip_stops.sort_values("stop_sequence_num")

    print("\nOriginal stop-time records:")
    display(trip_stops)
```

```python
import pandas as pd

# Convert GTFS time (HH:MM:SS) into total seconds
def gtfs_time_to_seconds(value):
    try:
        hours, minutes, seconds = map(int, value.split(":"))
        return hours * 3600 + minutes * 60 + seconds
    except (ValueError, AttributeError):
        return None

# Work with original Hyderabad stop-time data
duration_df = original_stop_times.copy()

duration_df["arrival_seconds"] = (
    duration_df["arrival_time"].apply(gtfs_time_to_seconds)
)

duration_df["departure_seconds"] = (
    duration_df["departure_time"].apply(gtfs_time_to_seconds)
)

# Sort stops within each trip
duration_df["stop_sequence_num"] = pd.to_numeric(
    duration_df["stop_sequence"],
    errors="coerce"
)

duration_df = duration_df.sort_values(
    ["trip_id", "stop_sequence_num"]
)

# Calculate first arrival and last departure for each trip
trip_duration = duration_df.groupby("trip_id").agg(
    first_arrival=("arrival_seconds", "first"),
    last_departure=("departure_seconds", "last"),
    stop_count=("stop_id", "count")
).reset_index()

trip_duration["duration_hours"] = (
    trip_duration["last_departure"] -
    trip_duration["first_arrival"]
) / 3600

print("Total trips checked:", len(trip_duration))
print("Trips longer than 24 hours:",
      (trip_duration["duration_hours"] > 24).sum())

print("\nLongest 15 trips:")
display(
    trip_duration.sort_values(
        "duration_hours", ascending=False
    ).head(15)
)
```

```python
# Get the five longest trip IDs
longest_trip_ids = (
    trip_duration
    .nlargest(5, "duration_hours")["trip_id"]
    .tolist()
)

# Add route information
longest_trip_details = trip_duration[
    trip_duration["trip_id"].isin(longest_trip_ids)
].merge(
    original_trips,
    on="trip_id",
    how="left"
)

print("Five longest trips with route information:")
display(
    longest_trip_details[
        [
            "trip_id",
            "route_id",
            "service_id",
            "stop_count",
            "duration_hours"
        ]
    ].sort_values("duration_hours", ascending=False)
)

# Inspect the stop-time records for each trip
for trip_id in longest_trip_ids:
    print(f"\n========== TRIP {trip_id} ==========")

    records = original_stop_times[
        original_stop_times["trip_id"] == trip_id
    ].copy()

    records["sequence_number"] = pd.to_numeric(
        records["stop_sequence"],
        errors="coerce"
    )

    records = records.sort_values("sequence_number")

    print("Number of stops:", len(records))
    print("First five stops:")
    display(records.head(5))

    print("Last five stops:")
    display(records.tail(5))
```

```python
# Check whether stop IDs repeat within each trip
repeat_stop_check = (
    original_stop_times
    .groupby("trip_id")
    .agg(
        total_stops=("stop_id", "count"),
        unique_stops=("stop_id", "nunique")
    )
    .reset_index()
)

repeat_stop_check["repeated_stop_count"] = (
    repeat_stop_check["total_stops"] -
    repeat_stop_check["unique_stops"]
)

# Join with trip duration
repeat_stop_check = repeat_stop_check.merge(
    trip_duration[["trip_id", "duration_hours"]],
    on="trip_id",
    how="left"
)

# Check only trips longer than 24 hours
long_trips_repeat_check = repeat_stop_check[
    repeat_stop_check["duration_hours"] > 24
]

print("Trips longer than 24 hours:",
      len(long_trips_repeat_check))

print("Long trips containing repeated stop IDs:",
      (long_trips_repeat_check["repeated_stop_count"] > 0).sum())

print("\nFive longest trips and their repeated-stop counts:")

display(
    long_trips_repeat_check
    .sort_values("duration_hours", ascending=False)
    .head(10)
)
```

### Feature Selection Explanation

The following features were selected for the initial route-level machine learning dataset:

**Numerical features:**
- `trip_count`: Represents the number of trips associated with a route.
- `distinct_stop_count`: Represents the number of distinct stops served by a route.
- `average_stops_per_trip`: Represents the average number of stops per trip.

**Categorical feature:**
- `city`: One-hot encoded into five columns: Bangalore, Delhi, Hyderabad, Mumbai and Pune.

**Excluded features:**
- `route_id`: Identifier, not a numerical measurement. It should be retained separately for route identification.
- `agency_id`: Identifier; not included in the initial feature set.
- `route_type`: Excluded because it has only one unique value in the selected dataset.
- `service_count`: Excluded because it is constant.
- Route names: Descriptive text not included in the initial numerical modelling dataset.

Routes with unavailable numerical statistics were excluded from this modelling dataset without changing the original data.


### Machine Learning Strategy

The current processed dataset is prepared for initial unsupervised learning and clustering analysis.

- No target variable is created because the available GTFS data does not provide verified passenger-demand or occupancy labels.
- The dataset may be used for exploratory clustering methods such as K-Means.
- StandardScaler was applied to the three numerical features.
- City indicator columns were retained as binary values.
- The clustering results will require interpretation and validation before making conclusions about transit performance.

A supervised 80:20 training/validation split is not applicable to this current unsupervised dataset.

For future supervised machine learning, a valid target must first be obtained. The data should then be split into training and validation sets, and preprocessing transformations should be fitted only on the training data to avoid data leakage.
