```python
from google.colab import files
from pathlib import Path
from IPython.display import display
import pandas as pd
import zipfile

uploaded = files.upload()

archives = [
    Path(name) for name in uploaded
    if name.lower().endswith(".zip")
]

if len(archives) != 1:
    raise ValueError("Upload exactly one Module 2 results ZIP.")

PROJECT = Path("/content/RouteRanker_Module4")
INPUT = PROJECT / "input"
OUTPUT = PROJECT / "output"

INPUT.mkdir(parents=True, exist_ok=True)
OUTPUT.mkdir(parents=True, exist_ok=True)

with zipfile.ZipFile(archives[0]) as archive:
    root = INPUT.resolve()

    for member in archive.infolist():
        target = (INPUT / member.filename).resolve()

        if target != root and root not in target.parents:
            raise ValueError("Unsafe archive path.")

    archive.extractall(INPUT)

matches = list(INPUT.rglob("vapi_route_profiles.csv"))

if len(matches) != 1:
    raise FileNotFoundError(
        "Expected one vapi_route_profiles.csv. "
        "Upload the Module 2 results ZIP."
    )

source_profiles = pd.read_csv(
    matches[0],
    dtype="string",
    keep_default_na=False,
    encoding="utf-8-sig"
)

print("Rows and columns:", source_profiles.shape)
print("Service dates:", source_profiles["service_date"].unique())

display(source_profiles.head())

print("\nCategory variation:")
for column in ["route_type", "day_type", "day_of_week"]:
    if column in source_profiles:
        print(column, ":", source_profiles[column].unique().tolist())

print("\nStep 1 completed. Source values preserved.")
```
